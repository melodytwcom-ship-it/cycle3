本筆記本的目標有兩個：

1. 固定本專題的研究問題  
2. 從原始資料表中擷取 `WhatIsYourSex` 與 `CurrentAlcoholUse` 兩個變數，輸出只包含這兩欄的新資料表

## 1. Research Question

### Two-Proportion z-test

Is the proportion of current alcohol use different between male and female students?

**中文：**  
YRBS 2007 資料中，目前有飲酒習慣的男女學生比例是否有所不同？

| 項目 | 內容 |
|------|------|
| Group variable（分組變數） | `WhatIsYourSex` |
| Response variable（反應變數） | `CurrentAlcoholUse` |
| Method（方法） | Two-proportion z-test |

**假設：**

- H₀：p_male = p_female（男女飲酒比例相同）  
- H₁：p_male ≠ p_female（男女飲酒比例不同）

```python
# 匯入套件
import pandas as pd

# 設定顯示格式
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

# 讀取原始資料
df = pd.read_csv("../data/raw/YRBS_2007.csv")

# 顯示原始資料基本資訊
print("原始資料維度：", df.shape)
print("\n原始欄位名稱：")
print(df.columns.tolist())
```

```python
# 指定本專題要使用的兩個變數
selected_columns = [
    "WhatIsYourSex",
    "CurrentAlcoholUse"
]

# 擷取指定欄位
df_selected = df[selected_columns].copy()

# 顯示前幾列
print("擷取後資料維度：", df_selected.shape)
print("\n前 5 列資料：")
display(df_selected.head())
```

```python
# 建立輸出路徑並輸出只含兩個變數的新資料表
output_path = "../data/processed/cycle3_selected_variables.csv"
df_selected.to_csv(output_path, index=False)

print(f"已輸出檔案：{output_path}")

# 再次確認輸出的資料表欄位
print("\n輸出資料表欄位：")
print(df_selected.columns.tolist())
```

## 2. Output

本筆記本輸出的檔案為：

- `../data/processed/cycle3_selected_variables.csv`

此檔案只包含後續分析需要的兩個變數：

| 變數名稱 | 角色 | 說明 |
|----------|------|------|
| `WhatIsYourSex` | Group variable | 學生性別（1 = Female, 2 = Male） |
| `CurrentAlcoholUse` | Response variable | 目前飲酒頻率（1 = 未飲酒，2–7 = 有飲酒） |
